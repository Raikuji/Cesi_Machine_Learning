# Livrable Bibliographie — Projet IA for HumanForYou

**Formation :** PGE A4 FISA INFO — Intelligence Artificielle (Machine Learning)  
**Projet :** Analyse de l'attrition des employés — HumanForYou  


## Introduction

Ce document recense l'ensemble des références académiques, techniques et éthiques mobilisées dans le cadre du projet d'analyse de l'attrition des employés de HumanForYou. Chaque source est annotée pour préciser son apport spécifique à notre démarche et son impact sur nos choix méthodologiques. Les sources sont classées en quatre thématiques conformément aux consignes du livrable.


## 1. Sources méthodologiques et théoriques

---

**[1] James, G., Witten, D., Hastie, T., & Tibshirani, R. (2021). *An Introduction to Statistical Learning with Applications in R* (2e éd.). Springer.**  
Licence : libre accès (PDF disponible sur le site des auteurs)  
*Apport :* Référence fondamentale pour la compréhension des modèles de régression logistique, des méthodes de validation croisée (train/test split) et des métriques d'évaluation (AUC-ROC, matrice de confusion). A guidé notre approche de la modélisation supervisée et le choix des métriques adaptées à un problème de classification binaire déséquilibrée.

---

**[2] Hastie, T., Tibshirani, R., & Friedman, J. (2009). *The Elements of Statistical Learning* (2e éd.). Springer.**  
Licence : libre accès (PDF disponible sur le site des auteurs)  
*Apport :* Référence approfondie sur les fondements mathématiques des algorithmes utilisés (K-Means, régression logistique, Random Forest). A permis de comprendre les hypothèses sous-jacentes à chaque modèle et de justifier nos choix de paramétrage (notamment `class_weight='balanced'` pour la gestion du déséquilibre de classes).

---

**[3] Géron, A. (2022). *Hands-On Machine Learning with Scikit-Learn, Keras, and TensorFlow* (3e éd.). O'Reilly Media.**  
Licence : ouvrage commercial (droits d'auteur O'Reilly)  
*Apport :* Guide pratique ayant orienté notre implémentation des pipelines de préparation des données, du train/test split stratifié et de l'optimisation des hyperparamètres via GridSearchCV. Particulièrement utile pour la structuration du notebook Jupyter.

---

**[4] MacQueen, J. (1967). Some Methods for Classification and Analysis of Multivariate Observations. *Proceedings of the 5th Berkeley Symposium on Mathematical Statistics and Probability*, 1, 281-297.**  
Licence : domaine public  
*Apport :* Article fondateur de l'algorithme K-Means utilisé pour notre analyse de clustering. A permis de justifier le recours à la méthode du coude (Elbow Method) et au score de silhouette pour la sélection du nombre optimal de clusters.


## 2. Sources sur les aspects techniques

---

**[5] Pedregosa, F., et al. (2011). Scikit-learn: Machine Learning in Python. *Journal of Machine Learning Research*, 12, 2825-2830.**  
Licence : BSD (open source)  
*Apport :* Publication de référence de la bibliothèque scikit-learn, utilisée pour l'ensemble de notre pipeline de modélisation (LogisticRegression, KMeans, train_test_split, métriques). Citée pour valider la fiabilité et la reproductibilité des implémentations utilisées.

---

**[6] McKinney, W. (2010). Data Structures for Statistical Computing in Python. *Proceedings of the 9th Python in Science Conference*, 51-56.**  
Licence : BSD (open source)  
*Apport :* Publication de référence de la bibliothèque Pandas, utilisée pour le chargement, la fusion et la manipulation des 5 fichiers CSV du projet. La fonction `pd.get_dummies()` utilisée pour l'encodage One-Hot des variables catégorielles est documentée dans cette bibliothèque.

---

**[7] Hunter, J. D. (2007). Matplotlib: A 2D Graphics Environment. *Computing in Science & Engineering*, 9(3), 90-95.**  
Licence : PSF (open source)  
*Apport :* Bibliothèque de visualisation utilisée pour l'ensemble des graphiques du projet (distributions, boxplots, courbes ROC, matrices de confusion). Complétée par Seaborn pour les visualisations statistiques avancées (heatmaps, KDE plots).

---

**[8] Bilogur, A. (2018). Missingno: a missing data visualization suite. *Journal of Open Source Software*, 3(22), 547.**  
Licence : MIT (open source)  
*Apport :* Bibliothèque utilisée pour la visualisation des valeurs manquantes lors de la phase de préparation des données. A permis d'identifier rapidement les colonnes incomplètes et de justifier la stratégie d'imputation retenue (médiane pour les variables numériques, mode pour les variables catégorielles).

---

**[9] Virtanen, P., et al. (2020). SciPy 1.0: Fundamental Algorithms for Scientific Computing in Python. *Nature Methods*, 17, 261-272.**  
Licence : BSD (open source)  
*Apport :* Bibliothèque utilisée pour les tests statistiques bivariés : test de Mann-Whitney U (comparaison de distributions numériques entre groupes Attrition Yes/No) et test du Chi-2 (indépendance entre variables catégorielles et attrition). Ces tests ont guidé la sélection des variables pertinentes pour la modélisation.


## 3. Sources éthiques et sociétales

---

**[10] Commission Européenne — Groupe d'experts de haut niveau sur l'intelligence artificielle. (2019). *Lignes directrices en matière d'éthique pour une IA digne de confiance*. Publications Office of the European Union.**  
Licence : libre accès (document officiel UE)  
*Apport :* Document fondateur de notre démarche éthique. Les 7 exigences qu'il définit (autonomie humaine, robustesse, confidentialité, transparence, non-discrimination, bien-être sociétal, responsabilité) ont structuré l'intégralité de notre livrable éthique et ont guidé nos décisions tout au long du projet.

---

**[11] Règlement (UE) 2016/679 du Parlement européen et du Conseil (RGPD). (2016). *Journal officiel de l'Union européenne*.**  
Licence : libre accès (texte de loi)  
*Apport :* Cadre légal de référence pour le traitement des données personnelles des employés. A motivé notre décision de supprimer l'`EmployeeID` avant modélisation et d'encadrer l'usage des données de badgeuse. Les articles 13 et 14 relatifs à l'information des personnes concernées ont été particulièrement mobilisés.

---

**[12] Barocas, S., Hardt, M., & Narayanan, A. (2023). *Fairness and Machine Learning: Limitations and Opportunities*. MIT Press.**  
Licence : libre accès (fairmlbook.org)  
*Apport :* Référence sur les biais algorithmiques et l'équité dans les systèmes de machine learning. A guidé notre analyse des risques de discrimination liés aux variables `Gender`, `Age` et `MaritalStatus`, et notre recommandation de vérifier l'équité du modèle entre sous-groupes.

---

**[13] Code du travail français — Article L1132-1. Principe de non-discrimination.**  
Licence : libre accès (Légifrance)  
*Apport :* Cadre légal encadrant l'usage des critères protégés (sexe, âge, origine, situation de famille) dans les décisions de gestion du personnel. A motivé notre recommandation explicite de ne pas utiliser `Gender` ou `MaritalStatus` comme critères de décision individuelle.


## 4. Sources spécifiques au projet

---

**[14] Choudhary, V. (2018). *HR Analytics Case Study Dataset*. Kaggle.**  
URL : https://www.kaggle.com/vjchoudhary7/hr-analytics-case-study  
Licence : CC BY-SA 4.0 (usage autorisé avec attribution)  
*Apport :* Source directe des données du projet. Le dataset comprend les 5 fichiers CSV utilisés : `general_data.csv`, `employee_survey_data.csv`, `manager_survey_data.csv`, `in_time.csv`, `out_time.csv`. L'exploration de la page Kaggle a également permis de s'inspirer d'approches analytiques d'autres contributeurs, que nous avons adaptées et justifiées selon notre propre démarche.

---

**[15] IBM Watson Analytics. (2015). *Employee Attrition & Performance — HR Sample Dataset*.**  
Licence : usage éducatif  
*Apport :* Dataset de référence dans le domaine de l'analyse RH par machine learning, dont le dataset Kaggle utilisé dans ce projet est inspiré. A permis de contextualiser les variables (JobLevel, StockOptionLevel, OverTime) et de comprendre les pratiques RH du secteur pharmaceutique en Inde.

---

**[16] Chitra, U., & Subramanian, B. (2013). Employee Retention Using Predictive Modeling in IT Industry. *ISCA Journal of Engineering Sciences*, 2(7), 27-30.**  
Licence : accès libre  
*Apport :* Étude de cas sur la prédiction de l'attrition dans un contexte industriel. A confirmé la pertinence de variables comme `JobSatisfaction`, `MonthlyIncome` et `YearsAtCompany` comme prédicteurs majeurs du turnover, en cohérence avec nos résultats d'analyse bivariée.

---

**[17] Punnoose, R., & Ajit, P. (2016). Prediction of Employee Turnover in Organizations Using Machine Learning Algorithms. *International Journal of Advanced Research in Artificial Intelligence*, 5(9).**  
Licence : accès libre  
*Apport :* Article comparant plusieurs algorithmes de classification (Logistic Regression, Random Forest, Decision Tree) sur un problème d'attrition RH similaire au nôtre. A justifié notre choix d'évaluer plusieurs modèles et de prioriser le recall comme métrique principale dans un contexte de classes déséquilibrées.


## Conclusion

Les sources mobilisées dans ce projet couvrent l'ensemble des dimensions de notre démarche : les fondements théoriques du machine learning, les outils techniques Python utilisés, le cadre éthique et légal applicable aux projets IA en RH, et les travaux antérieurs spécifiques à la problématique d'attrition. Chaque source a été sélectionnée pour sa pertinence directe avec nos choix et non comme référence générique.

L'ensemble des bibliothèques open source utilisées (scikit-learn, Pandas, Matplotlib, SciPy, Missingno) sont sous licences permissives (BSD, MIT, PSF) autorisant leur usage dans un cadre académique et commercial sans restriction.
